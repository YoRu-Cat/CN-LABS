## Question 1

```python
def final_grade(quiz, midterm, final):
    if quiz < 40 or midterm < 40 or final < 40:
        return "F"
    percentage = quiz * 0.2 + midterm * 0.3 + final * 0.5
    if percentage >= 90:
        return "A"
    elif percentage >= 80:
        return "B"
    elif percentage >= 70:
        return "C"
    elif percentage >= 60:
        return "D"
    else:
        return "F"
print(final_grade(95, 88, 92))
print(final_grade(95, 88, 35))
```

## Question 2

```python
def is_armstrong(n):
    total = 0
    temp = n
    while temp > 0:
        digit = temp % 10
        total += digit ** 3
        temp = temp // 10
    return total == n
def find_armstrong_ports(start, end):
    result = []
    for n in range(start, end + 1):
        if is_armstrong(n):
            result.append(n)
    return result
print(is_armstrong(153))
print(find_armstrong_ports(100, 999))
```

## Question 3

```python
def password_strength(password):
    special_chars = "!@#$%^&*"
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False
    for ch in password:
        if ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True
        elif ch.isdigit():
            has_digit = True
        elif ch in special_chars:
            has_special = True
    score = 0
    if has_upper:
        score += 1
    if has_lower:
        score += 1
    if has_digit:
        score += 1
    if has_special:
        score += 1
    if len(password) >= 8:
        score += 1
    if score <= 2:
        label = "Weak"
    elif score <= 4:
        label = "Moderate"
    else:
        label = "Strong"
    return (score, label)
print(password_strength("fast123"))
print(password_strength("FastNU@2026"))
print(password_strength("@MagiaNegra"))
```

## Question 4

```python
def is_valid_ip(ip):
    parts = ip.split(".")
    if len(parts) != 4:
        return False
    for part in parts:
        if len(part) == 0 or len(part) > 3:
            return False
        if not part.isdigit():
            return False
        if len(part) > 1 and part[0] == "0":
            return False
        if int(part) > 255:
            return False
    return True
print(is_valid_ip("192.168.1.1"))
print(is_valid_ip("256.100.50.0"))
print(is_valid_ip("192.168.01.1"))
print(is_valid_ip("192.168.1"))
```

## Question 5

```python
def is_palindrome(s):
    rev = ""
    for ch in s:
        if ch.isalnum():
            rev += ch.lower()
    return rev == rev[::-1]
print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("Race a Car"))
```

## Question 6

```python
def flatten(nested_list):
    result = []
    for item in nested_list:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result
print(flatten([[1, 2, 3], [4, 5]]))
print(flatten([[0, 4, 7], [3, 5], 11]))
```

## Question 7

```python
def unique_values(d):
    seen = []
    for value in d.values():
        if value not in seen:
            seen.append(value)
    return seen
print(unique_values({"a": 1, "b": 2, "c": 1, "d": 3, "e": 2, "f": 1}))
```

## Question 8

```python
def min_max_tuples(tuples_list):
    last_values = [t[-1] for t in tuples_list]
    min_val = min(last_values)
    max_val = max(last_values)
    min_tuples = [t for t in tuples_list if t[-1] == min_val]
    max_tuples = [t for t in tuples_list if t[-1] == max_val]
    return min_tuples, max_tuples
print(min_max_tuples([(1, 2), (3, 4), (5, 0)]))
print(min_max_tuples([(1, 2), (3, 4), (5, 0), (4, 4), (1, 0)]))
```

## Question 9

```python
def transpose(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    result = []
    for c in range(cols):
        new_row = []
        for r in range(rows):
            new_row.append(matrix[r][c])
        result.append(new_row)
    return result
print(transpose([[1, 2, 3], [4, 5, 6]]))
```
